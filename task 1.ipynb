```python
# 1.Write a program to check whether a student is eligible for admission based on marks in three subjects using comparison and logical operators.
a = int(input("a is:"))
b = int(input("b is:"))
c = int(input("c is:"))

if a >= 90 and b >= 90 and c >= 90:
    print (' A - grade ' )
elif a >= 80 and b >= 80 and c >= 80:
    print ('B - grade' )
elif a >= 70 and b >= 70 and c >= 70:
    print( 'c - grade' )
else:
      print('d - grade' )
```

```python
#2.Write a program to calculate an employee's bonus based on salary, experience, and performance rating using arithmetic, comparison, and logical operators.
salary = int(input(":"))
experience = float(input(":"))
performance_rating = float(input(":"))


if salary >= 30000 and experience >= 5 and performance_rating >= 4.5:
    print (' 20% Bonus ', salary * 0.20 )
elif salary >= 20000 and experience >= 2.5 and performance_rating >= 3.5:
    print (' 10% Bonus ', salary * 0.10 )
elif salary >= 10000 and experience >= 1 and performance_rating >= 2.5:
    print (' 5% Bonus ', salary * 0.5 )
else:
      print('500 Rs Bonus' )
```

```python
#3.Write a program to calculate the final shopping amount after applying discounts based on purchase amount, membership status, and festival offer.
purchase_amount = int(input('amount'))
membership_status = bool(float(input('membership')))
festival_offer = bool(int(input('festival')))

if purchase_amount >= 10000 and (membership_status or festival_offer):
    print('offer - 20%')
    
elif purchase_amount >= 5000 and (membership_status or festival_offer):
    print('offer - 10%')
    
elif purchase_amount >= 2000 and (membership_status or festival_offer):
    print('offer - 5%')

else :
    print('offer - 0%')
```
