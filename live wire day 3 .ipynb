```python
#Write a program to calculate the area and perimeter of a rectangle,square and circle.
# rectangle
length = 20
width = 10

rect_area = length * width
rect_perimeter = 2 *( length + width )

print("area:", rect_area )
print("perimeter:", rect_perimeter)

#square
length = 10
width = 5

square_area = length * width
square_perimeter = 4 *(length + width)

print("area:", square_area )
print("perimeter:", square_perimeter)

#circle

radius = 10
pi =3.14

circ_area = pi * radius * radius
circ_circumference = 2 * pi * radius

print("area:",circ_area)
print("circumference:",circ_circumference)
```

```python
'''4.Write a program to check:
i)If two numbers are equal
ii)If one number is greater than the other
iii)If a number is less than or equal to another '''
num1 = float(input("Enter the first number: "))
num2 = float(input("Enter the second number: "))
print("-" * 30)

if num1 == num2 :
    print("two numbers are equal")
else :
    print("two numbers are not equal")

if num1 > num2 :
    print("one number is greater than the other")
else :
    print("one number is not greater than the other")

if num1 <= num2 :
    print("number is less than or equal to another")
else :
    print("number is not less than or equal to another")
```

```python
# 1. Starting values
a = 10
b = 5

print("Before: a =", a)
print("Before: b =", b)
print("-" * 30)

# 2. The Math Steps
a = a + b    # a becomes 15
b = a - b    # b becomes 10 (15 - 5)
a = a - b    # a becomes 5  (15 - 10)

# 3. Final values
print("After: a =", a)
print("After: b =", b)
```
